#U1_A5 - ENCERRAMENTO


```python
def calcular_desconto(preco, percentual):
  if percentual < 0 or percentual > 100:
    return None

  desconto = preco * (percentual/100)
  return preco - desconto

def registrar_venda(produto, preco):
  print(f"\nVenda registrada: Produto: {produto} | Preço: {preco:.2f}")

arredondar = lambda valor: round(valor, 2)

total_vendas = 0
quantidade_vendas = 0
lista_produtos = []

while True:
  print("\n--- MeuPDV Python ---")
  print("1 - Iniciar Venda")
  print("2 - Relatório do dia")
  print("0 - Encerrar Caixa")
  opcao = int(input("Escolha uma opção: "))

  if opcao == 0:
   print("\nEncerrando caixa. Até logo!")
   break

  elif opcao == 1:
    print("\n--- Venda ---")
    produto = input("Nome do Produto: ")
    preco = float(input("Preço: R$ "))
    percentual = float(input("Desconto (%): "))
    preco_final = calcular_desconto(preco, percentual)

    if preco_final is None:
      print("Desconto inválido! Informe um valor ente 0 e 100")
    else:
      preco_final = arredondar(preco_final)
      registrar_venda(produto, preco_final)
      lista_produtos.append(produto)
      total_vendas = total_vendas + preco_final
      quantidade_vendas = quantidade_vendas + 1

  elif opcao == 2:
    print("\n--- Vendas do dia ---")
    print(f"Quantidade de Vendas: {quantidade_vendas}")
    print(f"Total Faturado: R$ {arredondar(total_vendas)}")

    print("\n--- Produtos Vendidos ---")
    if len(lista_produtos) == 0:
      print("Nenhum produto vendido até o momento.")
    else:
      for produto in lista_produtos:
        print(produto)

  else:
    print("Opção inválida! Certifique de escolher 1,2 ou 0")

```

#U2_A3 - Classes e Métodos em Python


```python
class Veiculo:
  def __init__(self, marca, modelo, ano, preco):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.preco = preco

  def resumo(self):
    return f"Modelo: {self.modelo}, Marca: {self.marca}, Ano: {self.ano}, Preço: {self.preco}"

class Carro(Veiculo):
  def __init__(self, marca, modelo, ano, preco):
    super().__init__(marca, modelo, ano, preco)

  def resumo(self):
    return super().resumo()

class Moto(Veiculo):
  def __init__(self, marca, modelo, ano, preco):
    super().__init__(marca, modelo, ano, preco)

  def resumo(self):
    return super().resumo()


carro1 = Carro("Chevrolet", "S10 High Country", 2024, 200.000)
moto1 = Moto("Yamaha", "Factor 125", 2014, 8.900)

print("Carros: ")
print(carro1.resumo())
print("\nMotos: ")
print(moto1.resumo())
```

#U2_Atividade Prática - Sistemas de gerenciamento de Livros


```python
#Importando a biblioteca de visualização de dados em gráficos
import matplotlib.pyplot as plt

#criando classe que representa um livro
class Livro:
  #definindo os atributos da classe
  def __init__(self, titulo, autor, genero, quantidade):
    self.titulo = titulo
    self.autor = autor
    self.genero = genero
    self.quantidade = quantidade
 #função para forma como os livros serão exibidos para o usuário
  def __str__(self):
    return (f"Título: {self.titulo} | Autor: {self.autor} | Genêro: {self.genero} | Quantidade: {self.quantidade}")

#criando a lista de livros
livros = []
#criando a lista de genêros
generos = []
#função para cadastrar livros
def adicionar_livro():
  print("\n--- Cadastro de Livro ---")
  titulo = input("Titulo do Livro: ")
  autor = input("Autor: ")
  genero = input("Genêro: ")
  quantidade = int(input("Quantidade: "))
  #criando variável para atribuir as entradas de usuário a classe
  novo_livro = Livro(titulo, autor, genero, quantidade)
  #salvando o livro na lista de livros
  livros.append(novo_livro)
  #salvando o genêro na lista de genêros
  generos.append(genero)
  #mensagem de sucesso
  print(f"Livro {titulo} cadastrado com sucesso!")

#função para listar livros cadastrados
def listar_livros():
  #verificando se a lista está vazias
  if len(livros) == 0:
    #se tiver vazia exibe a mensagem
    print("\nNenhum livro cadastrado.")
  #caso não esteja, lista os livros
  else:
    print("\n--- Livros ---")
    #percorre a lista de livros e exibe os cadastrados
    for livro in livros:
      print(livro)
#função para buscar livro pelo título
def buscar_titulo(livros, titulo_busca):
  #percorre a lista de livros
  for livro in livros:
    #verifica se o titulo é igual ao título informado
    if livro.titulo == titulo_busca:
      print("\nnLivro encontrado: ")
      return livro
 #caso não encontre, exibe mensagem
  print("\nLivro não encontrado!")
  return None

def livros_por_genero():
  if len(livros) == 0:
    print("Nenhum livro cadastrado ainda.")
    return

  #usando set para evitar genêros duplicados
  generos_unicos = list(set(generos))
  generos_unicos.sort()

  #Usando List Comprehesions para somar a quantidade de livros que pertence a cada gênero
  qtd_genero = [
      sum(livro.quantidade for livro in livros if livro.genero == g)
      for g in generos_unicos
  ]

  # Criando o gráfico de barras
  plt.bar(generos_unicos, qtd_genero, color='blue')
  # Títulos dos eixos x e y
  plt.xlabel('Gêneros')
  plt.ylabel('Quantidade Total de Exemplares')
  # Adicionando um título ao gráfico
  plt.title('Livros por Gênero')
  # Exibindo o gráfico
  plt.show()

while True:
  print("\n--- Gerenciador de Livros ---")
  print("\nMenu:")
  print("1 - Cadastrar livros")
  print("2 - Ver livros")
  print("3 - Buscar livro pelo título")
  print("4 - Livros por genêro")
  print("0 - Sair do sistema")
  opcao = int(input("Escolha uma opção: "))

  if opcao == 1:
    #chama função de adicionar livros
    adicionar_livro()
  elif opcao == 2:
    #chama função de listar livros
    listar_livros()
  elif opcao == 3:
    #pede para o usuário digitar o titulo que deseja buscar
    titulo_busca = input("Digite o título do livro: ")
    #chama função de buscar livro pelo título
    livro_encontrado = buscar_titulo(livros, titulo_busca)
    print(livro_encontrado)
  elif opcao == 4:
    #chama função que exibe o gráfico
    livros_por_genero()
  elif opcao == 0:
    #condição de encerramento do loop
    print("\nSaindo do sistema...")
    break
  else:
    #trata as entradas inválidas no menu
    print("Opção Inválida!")
```

#U3_A1 - Aplicação com banco de dados em Python

```python
import sqlite3

conexao = sqlite3.connect('contato.db')
cursor = conexao.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS contatos(
      id INTEGER PRIMARY KEY AUTOINCREMENT,
      nome TEXT NOT NULL,
      telefone TEXT NOT NULL,
      email TEXT NOT NULL
      )
""")

dados = [
    ('Anderson', '88123-4567', 'anderson@gmail.com'),
    ('Belmiro', '88988-7766', 'bel@gmail.com'),
    ('Bel', '88988-7766', 'bel@gmail.com')
]

cursor.executemany('INSERT INTO contatos(nome, telefone, email) VALUES(?,?,?)', dados)

conexao.commit()

cursor.execute('SELECT * FROM contatos')
contatos = cursor.fetchall()
print("Contatos:")
for contato in contatos:
  print(contato)

novo_telefone = '88923-4545'
id_contato = 1
cursor.execute('UPDATE contatos SET telefone = ? WHERE id = ?', (novo_telefone, id_contato))
conexao.commit()
nome_sql = "SELECT nome FROM contatos WHERE id = ?"
cursor.execute(nome_sql, (id_contato,))
resultado = cursor.fetchone()
nome_contato = resultado[0]
print(f"Alterando o telefone do contato {nome_contato} para {novo_telefone}")

id_contato_excluido = 3
cursor.execute('DELETE FROM contatos WHERE id = ?', (id_contato_excluido,))
nome_sql = "SELECT nome FROM contatos WHERE id = ?"
cursor.execute(nome_sql, (id_contato,))
contato_excluido = resultado[0]
print(f"Contato {contato_excluido} excluído com sucesso!")

conexao.close()

```

#U3_A3 - Introdução a Manipulação de dados em Pandas


```python
import pandas as pd

# Criando um DataFrame com 5 linhas de dados
data = {
'nome': ['Produto A', 'Produto B', 'Produto C', 'Produto A', 'Produto E'],
'quantidade de itens comprados': [3, 1, 4, 3, 2],
'tipo de item': ['Eletrônico', 'Vestuário', 'Alimento', 'Eletrônico', 'Alimento'],
'receita total': [120, 80, 60, 120, 90]
}

df = pd.DataFrame(data)

# Duplicando uma linha
df.drop_duplicates(keep='last', inplace=True)

# Calculando a coluna 'preço do item'
df['preço do item'] = df['receita total'] / df['quantidade de itens comprados']

# Selecionando preço do item acima de 50 reais
itens_acima_de_50 = df[df['preço do item'] > 50]
print(“Itens acima de 50 reais:”)
print(itens_acima_de_50)

```
